# Python Aggregation Assignment

```python
import pandas as pd 
import numpy as np
```

# Task 1 – The Regional Performance Analyzer

```python
df = pd.read_excel("16/sales.xlsx")
df.head()
```

```python
# Group by Region (COUNTRY)
result = (
    df.groupby("COUNTRY")
      .agg(
          Total_Sales=("SALES", "sum"),
          Average_Sales=("SALES", "mean"),
          Total_Quantity=("QUANTITYORDERED", "sum")
      )
      .reset_index()
      .sort_values(by="Total_Sales", ascending=False)
)

print(result)
```

# Task 2 – The Market Strategist

```python
# Group by Region and Product Category
result = (
    df.groupby(["COUNTRY", "PRODUCTLINE"])
      .agg(
          Total_Sales=("SALES", "sum"),
          Total_Quantity=("QUANTITYORDERED", "sum")
      )
      .reset_index()
      .sort_values(by=["COUNTRY", "Total_Sales"], 
                   ascending=[True, False])
)

print(result)
```

# Task 3 – The Advanced Aggregator

```python
# Group by Sales Channel (DEALSIZE)
result = (
    df.groupby("DEALSIZE")
      .agg(
          Total_Sales=("SALES", "sum"),
          Average_Sales=("SALES", "mean"),
          Max_Sale=("SALES", "max"),
          Min_Sale=("SALES", "min"),
          Total_Orders=("ORDERNUMBER", "count")
      )
      .reset_index()
)

print(result)
```

# Task 4 – The Sales Intelligence Matrix

```python
# Create Pivot Table
pivot_result = pd.pivot_table(
    df,
    values="SALES",
    index="PRODUCTLINE",      # Rows → Product_Category
    columns="COUNTRY",        # Columns → Region
    aggfunc="sum",            # Aggregation → sum
    margins=True,             # Grand totals
    margins_name="Grand_Total"
)

print(pivot_result)
```

# Task 5 – The Market Frequency Analyzer

```python
freq_table = pd.crosstab(
    df["DEALSIZE"],
    df["PRODUCTLINE"],
    margins=True,
    margins_name="Total"
)

print(freq_table)
```
