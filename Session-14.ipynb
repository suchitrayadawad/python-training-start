# Task 1: The Record Reconciler (Duplicate Identification & Resolution)

```python
customers = pd.read_csv("customers-1000.csv")

print("Dataset Shape:",customers.shape)
print("Columns:",customers.columns.tolist())

customers.head()
```

```python
# Create Customer_Email column
customers["Customer_Email"] = customers["Email"]


# Simulate duplicate email records
customers.loc[5, "Customer_Email"] = customers.loc[0, "Customer_Email"]
customers.loc[10, "Customer_Email"] = customers.loc[0, "Customer_Email"]


# Create an Updated_Date column
customers["Updated_Date"] = pd.date_range(
    start="2024-01-01",
    periods=len(customers),
    freq="D"
)


# Give the duplicate records newer dates
customers.loc[5, "Updated_Date"] = pd.Timestamp("2025-01-01")
customers.loc[10, "Updated_Date"] = pd.Timestamp("2025-02-01")

customers.head(11)
```

```python
# Identify all duplicate Customer_Email records
duplicates = customers[
    customers.duplicated(
        subset="Customer_Email",
        keep=False
    )
]

display(duplicates)
```

```python
# Sort by Updated_Date so the newest record comes last
customers_sorted = customers.sort_values("Updated_Date")


# Remove duplicate Customer_Email values, keeping the most recent record
clean_customers = customers_sorted.drop_duplicates(
    subset="Customer_Email",
    keep="last"
)

print("Original number of records:", len(customers))
print("Records after removing duplicates:", len(clean_customers))

display(clean_customers.head(11))
```

# Impact of keep parameter in drop_duplicates()

- `keep='first'`: Keeps the first occurrence of each duplicate record and removes the later duplicates.

- `keep='last'`: Keeps the last occurrence of each duplicate record and removes the earlier duplicates. In this task, sorting by `Updated_Date` first allows us to keep the most recent record.

- `keep=False`: Removes all records that have duplicates. It does not keep any occurrence of the duplicated records.

# Task 2: The Text Normalizer (String Standardization)

```python
import pandas as pd

products = pd.read_csv("products.csv")

print("Dataset Shape:", products.shape)
print("Columns:", products.columns.tolist())

products.head()
```

```python
# Convert ProductName to lowercase
products["ProductName"] = products["ProductName"].str.lower()

products.head()
```

```python
# Find products containing the word "pro"
pro_products = products[
    products["ProductName"].str.contains("pro", na=False)
]

display(pro_products)
```

```python
# Create Is_Premium_Product column
products["Is_Premium_Product"] = products["ProductName"].str.contains(
    "pro",
    na=False
)

display(products)
```

# Why case normalization is mandatory

Case normalization is important because the same product name can appear with different capitalization, such as "Pro", "PRO", or "pro". Without converting the text to a consistent case, searches and filters may miss matching products.

Converting ProductName to lowercase ensures that text searches such as "pro" work consistently across the entire dataset. This improves data consistency, searching, filtering, and analysis.

# Task 3: The Compliance Scanner (Regex Pattern Extraction)

```python
import pandas as pd

df = pd.DataFrame({
    "Support_Message": [
        "REF12345 user email john@example.com issue logged",
        "Ticket 67890 assigned to REF5678 contact jane@gmail.com",
        "REF99999 ticket 12345 email support@company.com"
    ]
})

df
```

```python
df["Ticket_ID"] = df["Support_Message"].str.findall(r"\d+")

df
```

```python
df["REF_Code"] = df["Support_Message"].str.findall(r"^REF\w+")

df
```

```python
df["Email"] = df["Support_Message"].str.findall(r"\w+@\w+\.\w+")

df
```

# Regex Pattern Extraction

- `\d+` extracts numeric values from the support messages.
- `^REF\w+` extracts reference codes beginning with `REF`.
- `\w+@\w+\.\w+` extracts email addresses.

The extracted values were stored in separate columns: `Ticket_ID`, `REF_Code`, and `Email`.

# Task 4: The Data Sanitizer (Hybrid Cleaning Pipeline)

```python
df.head()
```

```python
# Task 4: Data Sanitizer - Hybrid Cleaning Pipeline

# 1. Remove duplicate emails
df = df.drop_duplicates(subset="Email")

# 2. Convert text values to lowercase
def lowercase_values(x):
    if isinstance(x, str):
        return x.lower()
    elif isinstance(x, list):
        return [item.lower() if isinstance(item, str) else item for item in x]
    else:
        return x

text_cols = df.select_dtypes(include="object").columns
df[text_cols] = df[text_cols].apply(
    lambda col: col.map(lowercase_values)
)

# 3. Extract 10-digit phone numbers
import re

df["Phone_Number"] = df["Support_Message"].apply(
    lambda x: re.findall(r"\d{10}", x)
)

df
```

# Why order of operations matters

The order of cleaning operations is important because each step prepares the data for the next step. First, duplicate emails are removed so duplicate records do not affect the dataset. Next, text values are converted to lowercase to maintain consistent formatting. Finally, phone numbers are extracted using a regular expression from the cleaned text. Following a logical order helps produce consistent and model-ready data.

# Task 5: The Audit Validator (Quality Assurance Check)

```python
# 1. Count duplicates after cleaning
df_check = df.copy()

# Convert list values to tuples so pandas can compare them
for col in df_check.columns:
    df_check[col] = df_check[col].apply(
        lambda x: tuple(x) if isinstance(x, list) else x
    )


duplicate_count = df_check.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)


# 2. Verify lowercase transformation
lowercase_check = df["Support_Message"].apply(
    lambda x: x == x.lower()
)

print("All Support_Message values are lowercase:", lowercase_check.all())


# 3. Confirm no null Ticket IDs
null_ticket_ids = df["Ticket_ID"].isna().sum()

print("Number of null Ticket IDs:", null_ticket_ids)
```

# How these checks prevent downstream failures

These quality checks help ensure that the cleaned dataset is reliable before it is used for further analysis or loaded into a data warehouse. Checking for duplicates prevents repeated records from affecting analysis and reports. Verifying lowercase transformation ensures consistent text formatting and avoids mismatches during text processing. Confirming that Ticket_ID values are not null ensures that every record can be uniquely identified and tracked. These checks help prevent incorrect analysis, broken data pipelines, and unreliable downstream results.
